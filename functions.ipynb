Introduction to Function

```python
def odd_or_even(num):
    """This function finds odd or even"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
odd_or_even(26)
```

```python
odd_or_even(51)
```

```python
#function with multiple parameters
def add(a,b):
    return a+b

add(3,8)
```

```python
#Default parameters
def greet(name="Guest"):
    print(f"Hello {name}")

greet("Kritika")
greet()
```

```python
#Variable length arguments
#positional and keyword argument
#positional arguments
def print_numbers(*args):
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,"kritz")
```

```python
#keyword arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="kritika",age="20")
```

```python
def print_details(*args,**kwargs):
    for numbers in args:
        print(numbers)
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(2,3,4,5,"kritz",name="kritika",age="20")
```

```python
#return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(2,4)
```
