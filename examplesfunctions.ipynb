```python
#temperature conversion

def convert_temp(temp,unit):
    if unit=='C':
        return temp * 9/5 + 32
    elif unit=='F':
        return (temp-32)*5/9 
    else:
        return None

convert_temp(25,'C')
convert_temp(77,'F')
```

```python
#password strength checker
def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True



is_strong_password("WeakPwd")
```

```python
is_strong_password("Strong16Pwd*")
```

```python
#Calculate total Cost of items in a shopping cart

def calculate_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost +=item['price']*item['quantity']
        return total_cost
cart=[
    {'name':'kritika','price':0.3,'quantity':2},
    {'name':'Banana','price':0.9,'quantity':8},
    {'name':'watermelon','price':0.6,'quantity':7},
]
total_cost=calculate_total_cost(cart)





total_cost=calculate_total_cost(cart)
print(total_cost)
```

```python
#check if string is palindrome
def is_palindrome(s):
    s = s.lower().replace(" ","")
    return s==s[::-1]
is_palindrome("A man a plan a canal Panama")
```

```python
#factorials of a number using recursion

def factorial(n):
    if n==0:
        return 1
    else:
        return n * factorial(n-1)
    
factorial(6)
```
